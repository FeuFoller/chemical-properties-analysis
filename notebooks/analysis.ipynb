```python
plt.hist(df["Solubility "])
plt.title("Distribution of Measured Log Solubility")
plt.xlabel("Log Solubility")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Solubility"], bins=30)

plt.title("Distribution of Solubility")
plt.xlabel("Solubility")
plt.ylabel("Frequency")

plt.savefig("../figures/solubility_distribution.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["MolWt"], df["Solubility"])

plt.title("Molecular Weight vs Solubility")
plt.xlabel("Molecular Weight")
plt.ylabel("Solubility")

plt.savefig("../figures/molwt_vs_solubility.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["MolLogP"], df["Solubility"])

plt.title("LogP vs Solubility")
plt.xlabel("LogP")
plt.ylabel("Solubility")

plt.savefig("../figures/logp_vs_solubility.png", dpi=300)

plt.show()
```
